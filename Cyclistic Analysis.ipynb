```python
import pandas as pd
```

```python
#Loading cleaned data set
df = pd.read_csv("cleaned 2025 data.csv")

df.info()
```

"started_at","ended_at","and "ride_length" coulmns where converted back to str when exported to csv.

```python
#tunring ride length to a time delta for duration
df["ride_length"] = pd.to_timedelta(df["ride_length"])

#finding more details of when and how long riders ride for.
print(f"Average duration of ride:{(df["ride_length"].mean().floor("s"))}")
print(f"Max ride length:{df["ride_length"].max().floor("s")}")
print(f"Most biked days of the week:{df["day_of_week"].mode()}")

print(f"Count of rides per day of the week:{df["day_of_week"].value_counts().head(7)}")

print("\nAverge length of rides per day of the week")
df.groupby("day_of_week")["ride_length"].mean().dt.floor("s").sort_values(ascending=False)
```

```python
#cameback to randomly delete data. the file was 1.07gb and tablaeu public only take up 1.00gb
#deleted 8% of data. random state is to maintian consistency everytime the code is ran
df = df.sample(frac=0.92, random_state=42)

#finished analyzing and exporting to csv to visualize in tablaeu. 
df.to_csv("Analyzed 2025 data.csv", index=False)
```
