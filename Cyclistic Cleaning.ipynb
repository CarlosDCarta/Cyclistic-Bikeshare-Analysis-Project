```python
import pandas as pd
```

```python
#Importing and combining data
df1 = pd.read_csv(r"202501-divvy-tripdata.csv")
df2 = pd.read_csv(r"202502-divvy-tripdata.csv ")
df3 = pd.read_csv(r"202503-divvy-tripdata.csv")
df4 = pd.read_csv(r"202504-divvy-tripdata.csv")
df5 = pd.read_csv(r"202505-divvy-tripdata.csv")
df6 = pd.read_csv(r"202506-divvy-tripdata.csv")
df7 = pd.read_csv(r"202507-divvy-tripdata.csv")
df8 = pd.read_csv(r"202508-divvy-tripdata.csv")
df9 = pd.read_csv(r"202509-divvy-tripdata.csv")
df10 = pd.read_csv(r"202510-divvy-tripdata.csv")
df11 = pd.read_csv(r"202511-divvy-tripdata.csv")
df12 = pd.read_csv(r"202512-divvy-tripdata.csv")
```

```python
# checking sizes of each data frame and if anything sticks out. 
df_list = [df1,df2,df3,df4,df5,df6,df7,df8,df9,df10,df11,df12]
n = 0
for i in df_list:
    n += 1
    print(f"\n df{n}-------------------------------------------------------------------------------------")
    print(f"{i.info()}")
n = 0
```

```python
df = pd.concat([df1,df2,df3,df4,df5,df6,df7,df8,df9,df10,df11,df12])
df
```

checking data frame and cleaning

```python
#checking for null values

print("\nMissing values for df:")
print(df.isnull().sum())
```

```python
#chekcing for duplicates values

print("\nDuplicate values for df:")
print(df.duplicated().sum())
```

```python
#finding distribution of df's missing values
for i in df_list:
    n += 1
    print(f"df{n}------------------------------------------------------------------------------------------")
    print(i.isnull().sum())
    print(f"df{n} WITH values ------------------------------------------------------------------------------------------")
    print(i.notnull().sum())
n = 0
```

Decided not drop rows with null's since they still had usful data. 
Like making a column form the "started_at" and "ended_at" columns. 

```python
#Changing "started_at" and "ended_at" columns to Date data type from str

df["started_at"] = pd.to_datetime(df["started_at"])
df["ended_at"] = pd.to_datetime(df["ended_at"])
df.info()
print(f"\n---------------------------------------------------------------------------------------------------------------------------")
df.head(20)
```

```python
#Adding Ride length column for every row. 
df["ride_length"] = (df["ended_at"] - df["started_at"]).dt.floor("s")
df
```

```python
import datetime
```

```python
#added day of the week coulmn and month to data frame
df["day_of_week"] = df["started_at"].dt.strftime("%A")
df["mont_of_ride"] = df["started_at"].dt.strftime("%B")
df.head(20)
```

```python
#exporting to analysis notebook
df.to_csv("cleaned 2025 data.csv", index=False)
```
